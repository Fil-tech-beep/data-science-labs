You have:
- A bunch of documents (news), each is just text (strings).

Clustering algorithms (k-means, DBSCAN, etc.) want:
- Numbers, usually: a matrix X of shape (n_documents, n_features).

So you must do:  
> text  →  numbers

> That step is called data preparation / weighting schema.

⸻

**“Weighting schema” = how much each word matters**

Idea:  
For each document you choose a vocabulary (set of words), and you give a weight to each word in each document.  

Simplest example: bag-of-words (counts)  
- Features: each column = a word (e.g. football, computer, pizza…)
- For each document: value = how many times that word appears.

> Better example (what they want): **TF-IDF**
- TF = Term Frequency: “How often does the word appear in this document?”
- IDF = Inverse Document Frequency: “How rare is this word across all documents?”
    - If a word appears in almost all docs (e.g. “the”, “and”), it gets down-weighted.
	- If it appears in few docs (“neural”, “goalkeeper”), it gets up-weighted.

Then:  
> TF-IDF weight = TF × IDF

So:  
- Words that are frequent in the document but rare overall get high weight → important for that doc.
- Words that are everywhere get low weight → mostly noise for clustering.



⸻

**What you practically do**

Pipeline in your head:  
1.	Load all texts → list of strings texts = [doc1, doc2, ..., docN]  

2.	Choose weighting schema:  
	- For this lab they basically say: “just use TF-IDF”.

3.	Apply TF-IDF:  
	- Conceptually: build a vocabulary, compute TF-IDF for each (doc, word).
	- You get a matrix X_tfidf with shape (n_docs, n_words).

4.	Feed X_tfidf to your clustering algorithm (k-means etc.).  

They’re not asking you to invent TF-IDF from scratch, just to understand that:  
> text → TF-IDF matrix → clustering

#### Nice we implemented a very inefficient TF and IDF computing, now compute TF–IDF:

For one document:
$\text{tfidf}(w, d) = \text{tf}(w, d) \cdot \text{idf}(w)$  

If your vocabulary is:  
$[\text{cat}, \text{dog}, \text{sat}, \text{on}, \text{mat}]$  

then Doc 1 becomes the vector:  
$\big[ \text{tfidf}(\text{cat}, d_1),\ \text{tfidf}(\text{dog}, d_1),\ \text{tfidf}(\text{sat}, d_1),\ \text{tfidf}(\text{on}, d_1),\ \text{tfidf}(\text{mat}, d_1) \big]$  

So something like:  
$[0.2\log(3/2),\ 0,\ 0.2\log(3/2),\ 0.2\log(3/2),\ 0.2\log(3)]$  

Doc 2 and Doc 3 would have their own vectors.  
Then clustering (KMeans, etc.) works on these numeric vectors.  



> TF changes per document  
> IDF is global

````python
# iterate over documents
# iterate over UNIQUE words inside one document,
# !! NO, NO NEED FOR THAT I ALREADY DID THAT --> ITERATE OVER THE LIST OF TFs, consider the firs TF !!
# consider each word inside the first TF
# compute TF-IDF by simply taking the TF value of the considered word * IDF value associated to that word
# this TF-IDF value is associated to a single word inside a single document, so cumulate the TF-IDFs among all documents in a list

TF_IDF_cumulate = []
for current_doc_TF in cumulate_TF:
    TF_IDF = {}
    for word in current_doc_TF:
        # current_doc_TF[word] TF value associated to this word in this document
        # IDF[word] IDF value associated to this word, in general, the IDF is global
        TF_IDF[word] = current_doc_TF[word] * IDF[word]
    TF_IDF_cumulate.append(TF_IDF)
print(TF_IDF_cumulate)

````

````python
import math

# -------------------------------------------------------------------------
# 0. Pick which documents you work on (now a subset, later you can use texts)
# -------------------------------------------------------------------------
docs = texts[:100]            # for debugging use texts[:2]

# -------------------------------------------------------------------------
# 1. Tokenize all documents
#    -> all_words_per_doc: list of lists, one list of words per document
#    -> all_words_global:  flat list of ALL words in ALL docs
# -------------------------------------------------------------------------
all_words_per_doc = []
all_words_global = []

for text in docs:
    words_per_doc = []
    lines = text.split('\n')
    for line in lines:
        words = line.split(' ')
        for word in words:
            word = word.strip()
            word = word.strip('():><|&#"".,;-()[]%$£!\t')
            if word != '':
                words_per_doc.append(word)
                all_words_global.append(word)
    all_words_per_doc.append(words_per_doc)

# -------------------------------------------------------------------------
# 2. Raw counts per document: cumulate_count_words_one_doc
#    -> one dict per doc: {word: count in that doc}
# -------------------------------------------------------------------------
cumulate_count_words_one_doc = []

for words_per_doc in all_words_per_doc:
    count_words_one_doc = {}
    for word in words_per_doc:
        if word in count_words_one_doc:
            count_words_one_doc[word] = count_words_one_doc[word] + 1
        else:
            count_words_one_doc[word] = 1
    cumulate_count_words_one_doc.append(count_words_one_doc)

# -------------------------------------------------------------------------
# 3. TF per document: cumulate_TF
#    -> one dict per doc: {word: tf(word, doc)}
#       tf(word, doc) = count(word, doc) / total_words_in_doc
# -------------------------------------------------------------------------
cumulate_TF = []

for count_words_one_doc in cumulate_count_words_one_doc:
    TF = {}
    # total number of words in this doc (sum of counts)
    N = 0
    for w in count_words_one_doc:
        N = N + count_words_one_doc[w]

    for word in count_words_one_doc:
        tf_value = count_words_one_doc[word] / N
        TF[word] = tf_value

    cumulate_TF.append(TF)

# -------------------------------------------------------------------------
# 4. DF and IDF
#    DF(word) = in how many documents does this word appear?
#    IDF(word) = log( N_docs / DF(word) )
# -------------------------------------------------------------------------
all_unique_words = set(all_words_global)

DF = {}
for unique_word in all_unique_words:
    # count how many docs contain this word at least once
    for words_one_doc in all_words_per_doc:
        if unique_word in words_one_doc:
            if unique_word in DF:
                DF[unique_word] = DF[unique_word] + 1
            else:
                DF[unique_word] = 1


####################### THIS IS THE CRITICAL PART, WOULD BE MORE EFFICIENT TO DO THIS: #######################
# for words_one_doc in all_words_per_doc:     # loop over documents
#     unique_words_in_doc = set(words_one_doc)  # words that appear in THIS doc at least once
    
#     for w in unique_words_in_doc:
#         ...


IDF = {}
N_docs = len(docs)       # here docs = texts[:2] for now
for word in DF:
    IDF[word] = math.log(N_docs / DF[word])

# -------------------------------------------------------------------------
# 5. TF–IDF per document
#    TF_IDF_cumulate: list of dicts, one dict per doc: {word: tfidf(word, doc)}
#    tfidf(word, doc) = tf(word, doc) * idf(word)
# -------------------------------------------------------------------------
TF_IDF_cumulate = []

for current_doc_TF in cumulate_TF:
    TF_IDF = {}
    for word in current_doc_TF:
        # safety: check word exists in IDF (it should, given how we built things)
        if word in IDF:
            TF_IDF[word] = current_doc_TF[word] * IDF[word]
    TF_IDF_cumulate.append(TF_IDF)

for idx in range(2):
    print(f'----------------------------------------------------- [DEBUG] The text: \n{texts[idx]}')
    print()
    print(f"----------------------------------------------------- [DEBUG] Has become this: {TF_IDF_cumulate[idx].values()}\n")
    print()

print(TF_IDF_cumulate)
````
