# REGRESSION THEORY
#### 1. Linear Regression  

Linear regression is a statistical method used for modeling the relationship between a dependent variable (y) and one or more independent variables (X). The goal is to find the line (or hyperplane in higher dimensions) that best fits the data, minimizing the sum of squared residuals.  

    - Formula:
$y = \beta_0 + \beta_1X_1 + \beta_2X_2 + ... + \beta_nX_n$

Where $\beta_0$ is the intercept, and $\beta_1$, $\dots$, $\beta_n$ are the coefficients of the independent variables $X_1, X_2, \dots, X_n$.

#### In linear regression, our model learns from the training data and basically draws a line. When we test it, like what are we doing? We’re feeding the model a bunch of points and what does it do?

In linear regression, during the training phase, the model learns the relationship between the features and the target by finding the best-fitting line (or hyperplane for higher dimensions). This is done by minimizing the error (usually the Mean Squared Error - MSE) between the predicted values and the actual values in the training set.

When we test the model:  
- We’re feeding the model new data that it has not seen before.
- The model takes these new input features (the points) and uses the learned coefficients (slope and intercept) to calculate the predicted output (predicted target values).
- The model then compares the predicted values with the true values (the labels) from the test set to measure its performance (accuracy, error, etc.).

> So, when testing, the model applies the learned coefficients to new data points to make predictions.

#### Why would I use linear regression? What for?
Linear regression is typically used for predictive modeling tasks where you want to predict a continuous outcome variable (also called the dependent variable or target variable) based on one or more independent variables.  

1. Predicting a Continuous Outcome (Prediction)  
- Example: Predicting the price of a house based on features like the size, number of rooms, location, and age of the house.
- Why Use Linear Regression: You want to model how the price of the house changes based on the values of the features. Linear regression will give you a formula to make predictions.

2. Assessing Relationships Between Variables  
- Example: Analyzing how a company’s advertising budget impacts its sales.
- Why Use Linear Regression: It helps understand the relationship between the independent variable (advertising budget) and the dependent variable (sales), and gives you an equation showing how sales change as advertising changes.

3. Trend Analysis (Time Series Forecasting)  
- Example: Predicting stock prices or economic indicators based on historical data.
- Why Use Linear Regression: You can use linear regression to detect trends in the data and make future predictions assuming a linear trend over time.

When NOT to Use Linear Regression:  
- When the relationship is not linear: If your features and target variable have a non-linear relationship, linear regression will not perform well.
- When there is high multicollinearity: If the independent variables are highly correlated, it can cause instability in the regression coefficients.
- When residuals are not normally distributed: Linear regression assumes that the errors (residuals) are normally distributed, so if this assumption is violated, the model may not be reliable.

Use linear regression when you:  
- Want to predict a continuous outcome.
- Need a simple model to understand relationships between variables.
- Have linearly related features with the target variable.

#### 2. MSE (Mean Squared Error)
MSE is a common metric for evaluating the performance of regression models. It calculates the average of the squared differences between actual and predicted values. Smaller values indicate better performance.  

    - Formula:  
$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

Where $y_i$ is the actual value, $\hat{y}_i$ is the predicted value, and $n$ is the number of data points.

#### SSE (Sum of Squared Errors)
It's just the numerator.  
SSE (Sum of Squared Errors) is related to MSE but without averaging by the number of data points.  
It calculates the sum of the squared differences between the actual and predicted values. SSE is often used when you want to calculate the total error in a model without averaging it over the data points.
- Formula for SSE:

$\text{SSE} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

#### 3. Polynomial Regression
Polynomial regression is an extension of linear regression where the relationship between the independent variable and the dependent variable is modeled as an n-th degree polynomial.  

    - Formula:
$y = \beta_0 + \beta_1X + \beta_2X^2 + ... + \beta_nX^n$

This allows the model to capture non-linear relationships.

#### 6. RMSE (Root Mean Squared Error)
RMSE is the square root of the MSE. It is more interpretable because it is in the same unit as the dependent variable, unlike MSE, which can be difficult to interpret due to its squared units.
    - Formula:
$RMSE = \sqrt{MSE}$

#### 7. MAE (Mean Absolute Error)
MAE is another evaluation metric that computes the average of the absolute differences between the actual and predicted values. Unlike MSE, it does not penalize large errors as severely.
    - Formula:
$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$

#### 8. R^2 (Coefficient of Determination)
R^2 is a metric that shows how well the regression model explains the variance in the dependent variable. It ranges from 0 to 1, where 1 indicates a perfect fit.
    - Formula:
$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2} = 1 - \frac{MSE}{variance^2}$

Where $\bar{y}$ is the mean of the actual values.

#### 9. Regularization: Overfitting, Underfitting
- Overfitting occurs when the model learns not only the underlying pattern in the training data but also the noise, leading to poor generalization on new data --> usually this creates a crazy, high-complexity, high-dimensionality model.

- Underfitting occurs when the model is too simple to capture the underlying patterns in the data, leading to poor performance on both training and test data --> dude draws a line and says that's enough.

Regularization techniques like **Lasso** and **Ridge** help prevent overfitting by adding a penalty to the model’s complexity.

#### Lasso (Least Absolute Shrinkage and Selection Operator)
Lasso is a regularization technique that is used to prevent overfitting in regression models by adding a penalty to the cost function based on the absolute values of the coefficients (weights). The Lasso regularization has a feature selection effect because it can drive some coefficients to zero, effectively eliminating some features from the model.  

    - Formula:
$\text{Cost} = \text{MSE} + \lambda \sum_{j=1}^{n} |\beta_j|$

Where:  
- $\lambda$ is the regularization strength (hyperparameter).
- $\beta_j$ represents the coefficients (weights) for each feature.
- MSE (Mean Squared Error) is used to measure the error between the model’s predictions and the true values.

How Lasso Works:  
- MSE measures how well the model fits the data, i.e., the difference between the predicted and actual values.
- The L1 penalty ($\sum |\beta_j|$) forces the model to shrink the coefficients of less important features toward zero.
- Larger coefficients are heavily penalized because their absolute values are larger.
- Smaller coefficients are penalized less.
- Feature Selection: Lasso can set certain coefficients to exactly zero, meaning those features are removed from the model. It effectively chooses the most important features.

Effect of Regularization Parameter ($\lambda$):
- Low $\lambda$: The regularization penalty is weak, so the model focuses on minimizing MSE. This can lead to overfitting, as many features are kept in the model.
- High $\lambda$: The penalty is stronger, forcing many coefficients to shrink toward zero. Features that contribute less to reducing the error are eliminated, leading to underfitting.

> Summary: Lasso is useful when you want to both prevent overfitting and automatically perform feature selection by shrinking some coefficients to zero.

#### Ridge Regression
Ridge regression is another regularization technique that also adds a penalty to the cost function, but instead of penalizing the absolute values of the coefficients (like Lasso), it penalizes the squared values of the coefficients. This helps to prevent large weights and makes the model more stable. 

    - Formula:

$\text{Cost} = \text{MSE} + \lambda \sum_{j=1}^{n} \beta_j^2$

Where:
- $\lambda$ is the regularization strength (hyperparameter).
- $\beta_j$ represents the coefficients (weights) for each feature.

How Ridge Works:  
- MSE is the same as in Lasso: it measures how well the model fits the data.
- The L2 penalty ($\sum \beta_j^2$) penalizes the square of the coefficients, which means that large coefficients are heavily penalized.
- Squaring the coefficients means that larger coefficients receive a stronger penalty while smaller penalties becomes even smaller, BUT NOT TO ZERO.
- Unlike Lasso, Ridge shrinks all coefficients, but none of them become exactly zero.
- Ridge forces the model to shrink all coefficients but does not eliminate any feature entirely.

Effect of Regularization Parameter ($\lambda$):  
- Low $\lambda$: The model behaves like regular linear regression, where the main goal is to minimize MSE. The coefficients of some features may be large.
- High $\lambda$: The penalty on large coefficients becomes more significant, forcing all coefficients to shrink. However, none of the coefficients become exactly zero.

Summary: Ridge is useful when you want to prevent overfitting and shrink coefficients, but you do not want to eliminate any features entirely.

#### 13. PolynomialFeatures()
The PolynomialFeatures class in scikit-learn is used to generate polynomial features from the original features of the dataset. This is useful when you’re trying to model nonlinear relationships, such as in polynomial regression.

What does it do?  
- PolynomialFeatures creates higher-degree polynomials of the original features. For example:
- If the input feature is X, it will create features like X^2, X^3, etc., depending on the degree you specify.

For example, suppose you have one feature X, and you want to generate features up to degree 2. The transformation would create two new features:
- X (the original feature),
- X^2 (the squared term).

**Example:**  
Let’s assume we have a dataset with a single feature (a simple column of numbers), like this:
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
```python
# Original feature: a simple column of numbers
X = np.array([[1], [2], [3], [4]])

# Create PolynomialFeatures object (degree=2 means we want up to X^2)
poly = PolynomialFeatures(degree=2)

# Transform the data
X_poly = poly.fit_transform(X)

print(X_poly)
```

- output:
```text
[[ 1.  1.  1.]
 [ 1.  2.  4.]
 [ 1.  3.  9.]
 [ 1.  4. 16.]]
```

- The first column 1.0 is the bias term (it’s always 1, representing X^0). --> it's the intercept with the y axis: y = mx + **q**
- The second column is the original feature X.
- The third column is the squared term X^2.

So, for each value of X in the original dataset, we now have X and X^2 as new features.

#### 14. fit_transform()
fit(): When you “fit” a transformer (like PolynomialFeatures), it learns the transformation from the data. In the case of PolynomialFeatures, this means:
- It learns the degree of the polynomial you want (this is set by the degree parameter).
- It understands how many features (columns) will be generated after applying the polynomial transformation.
- transform(): After the transformer is fitted, the transform step applies the transformation to the data, which results in a new dataset. For PolynomialFeatures, it takes the original features (e.g., X) and generates polynomial features like X^2, X^3, etc.
- fit_transform(): This method combines the two steps (fitting and transforming) into one, which is why it’s a more convenient option if you don’t need to call them separately. It first fits the transformer to the data (learns the polynomial degree) and then transforms the original data by applying the polynomial transformation.

```python
from sklearn.preprocessing import PolynomialFeatures
import numpy as np

# Original data: let's say we have one feature (X)
X = np.array([[1], [2], [3], [4]])

# Create PolynomialFeatures object (degree=2 means we want X^2)
poly = PolynomialFeatures(degree=2)

# Fit the model and transform the data (it learns the degree and then applies the transformation)
X_poly = poly.fit_transform(X)

print(X_poly)
```

- output:
```text
[[1. 1. 1.]
 [1. 2. 4.]
 [1. 3. 9.]
 [1. 4. 16.]]
```

- The first column (1.) is the bias term.
- The second column is the original feature X.
- The third column is the squared feature X^2.
- fit(): The PolynomialFeatures(degree=2) learns that we want to generate up to the second degree (i.e., X and X^2).
- transform(): After fitting, it transforms the original data into new polynomial features X^1 and X^2.
- fit_transform(): This method combines both steps and returns the transformed data.

#### 17. GridSearchCV()
