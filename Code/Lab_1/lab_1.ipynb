Sorting and ranking: Create a small integer matrix and apply np.sort along rows and columns. Then use np.argsort to see how the result changes.

### Sorting
- np.sort(x): creates a sorted copy of x
--> x is NOT modified

- x.sort(): sorts x inplace
--> x is modified

### Argsorting
- np.argsort(x): return the position of the indices of the sorted array (sorts by default on axis -1)

```python
m1 = np.array([[1.5, 2.6, -9.6],
               [0.5, 7.4, 8.2]])
print(m1)

print()

# Array is sorted along the last axis (-1) --> in a 2D matrix the last axis, so axis = 1, are the rows
sorted = np.sort(m1)    #sorts per rows
print(sorted)

print()

sorted_1 = np.sort(m1, axis=0)  #sorts per column
print(sorted_1)

print()

sorted_2 = np.sort(m1, axis=1)  #sorts per rows, it's the same as before
print(sorted_2)

print('------------------------- \n')

arg_sorted = np.argsort(m1)
# di default prende axis = -1 --> considera axis = 1 --> rows
# considera i valori [1.5, 2.6, -9.6] --> se li riordiniamo diventa [-9.6, 1.5, 2.6] --> l'argsort prende gli indici originali [2, 0, 1]
print()
print(arg_sorted)

arg_sorted_1 = np.argsort(m1, axis = 0)
# considers [1.5, 0.5], [2.6, 7.4], [-9.6, 8.2] --> quindi se li riordiniamo in base agli indici originali ottengo [1, 0], [0,1], [0,1]
# mettili tutti "di fianco":
# 1     0       0
# 0     1       1
print()
print(arg_sorted_1)
```

## Broadcasting
Broadcasting allows you to perform operations between arrays of different shapes without explicit loops. As an exercise, perform a column-wise normalization of a matrix by subtracting the minimum and dividing by the range for each feature.
(D−Dmin)/(Dmax−Dmin)

### Huh?
You have a **matrix** where each **column** is a feature (e.g., height, weight, age).  
**Column-wise normalization** means you make each column range from **0 to 1** by:

$ \text{Normalized} = \frac{D - D_{min}}{D_{max} - D_{min}} $

- \( D \) → the original data  
- \( D_{min} \) → the smallest value **in each column**  
- \( D_{max} \) → the largest value **in each column**  
- \( D_{max} - D_{min} \) → the **range** of that column  

You do this **for every column separately**.

```python
#example of broadcasting

m1 = np.array([[1, 2, 3],
               [4, 5, 6]])

broadcast_matrix = np.array([1, 2, 3])

addition = m1 + broadcast_matrix
print(addition)

print('----------------------')
example_matrix = np.array([[1.6, 8.4, -12.5, 11.1, 2.3],
                           [0.5, 7.4, 8.2, 0.1, 2.1],
                           [0.6, 7.5, 8.3, 0.2, 2.2]])

min_val = np.min(example_matrix, axis = 0)
max_val = np.max(example_matrix, axis = 0)
D_min_max = max_val - min_val

# So che la prima cosa che ti viene da fare è entrare in un ciclo for per far scorrere gli elementi, MA NO, puoi fare i calcoli tra matrici direttamente
# --> ecco perché questo esercizio è messo sotto broadcasting, prendi la matrice originale 3x5 e le sottrai un unico valore, il minimo --> broadcasting

normalized = (example_matrix - min_val) / D_min_max 
print(normalized)

```

### Slicing: access contiguous elements
- x[start:stop:step, ...]
- Creates a view of the elements from start (included) to stop (excluded), taken with fixed step
- Updates on the view yield updates on the original array

Useful shortcuts:
- omit start if you want to start from the beginning of the array

- omit step if you don't want to skip elements
