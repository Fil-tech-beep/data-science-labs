### Flight Delay Data

**Overview:**

This dataset comes from the **U.S. Department of Transportation** and tracks how well different airlines performed in January 2017, focusing specifically on **flight delays**.

Each row in the data represents **one flight** on a specific day. Here are the most important columns:

- **FL_DATE**: The date of the flight (YYYY-mm-dd).

- **UNIQUE_CARRIER**: The airline code.
- **FL_NUM**: The flight number.

- **WEATHER_DELAY**: Delay caused by the weather.
- **NAS_DELAY**: Delay caused by problems with the air traffic system.
- **SECURITY_DELAY**: Delay due to security checks.
